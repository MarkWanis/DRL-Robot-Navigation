<h3>Setting Up ROS2 Communication</h3>
<pre><code># Source the ROS 2 environment
source /opt/ros/lyrical/setup.bash
# Set the ROS domain ID and TurtleBot3 model for the system
export ROS_DOMAIN_ID={your_domain_id}
export TURTLEBOT3_MODEL=burger</pre></code>

<h4>Optional Configuration:</h4>
<pre><code># It is assumed these will be entered into your terminal and will not be repeated for the remaining tutorial
echo ‘source /opt/ros/lyrical/setup.bash’ >> ~/.bashrc
echo "export ROS_DOMAIN_ID={your_domain_id}" >> ~/.bashrc # Changes depending on which turtlebot you use
echo "export TURTLEBOT3_MODEL=burger" >> ~/.bashrc</pre></code>

<h4>Test Communication:</h4>

<h5>Listing Environment Variables</h5>

<h3>Creating Workspaces</h3>
<h4>Create a new workspace (i.e. root name = 'colcon_ws'):</h4>
<pre><code>mkdir -p ~/{workspace_name}/src # Ex: colcon_ws
cd ~/colcon_ws/src
# Make sure to run this line for each new project you create
ros2 pkg create --build-type ament_python {package_name} # Ex: drl_nav_project
cd ~/colcon_ws
rosdep install -i --from-path src --rosdistro {distro} -y # Choose your selected distributor: humble or lyrical
# Run this command every time you make a change to your project
colcon build
source install/local_setup.bash</code></pre>
